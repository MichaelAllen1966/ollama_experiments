# Running local LLMs in Python

## To install ollama and models

From terminal:

`curl -fsSL https://ollama.com/install.sh | sh`

`ollama pull gemma4:e4b` # Example model

`pip install openai ollama`

```python
from openai import OpenAI

client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama",  # required by the client, ignored by Ollama
)

response = client.chat.completions.create(
    model="gemma4:e4b",
    messages=[
        {"role": "system", "content": "You are a concise assistant with strong statistics knowledge."},
        {"role": "user", "content": "Explain Fast Causal Inference (FCI) in plain English."}
    ],
    temperature=0.3,
)

print(response.choices[0].message.content)
```

## Keeping model open in memory

```python
import os
from openai import OpenAI

client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama",
)
model_name = "gemma4:e4b"
```

```python
resp1 = client.chat.completions.create(
    model=model_name,
    messages=[{"role": "user", "content": "Give me one sentence about Newton Abbot."}],
    extra_body={"keep_alive": "30m"},
)

print(resp1.choices[0].message.content)
```

```python
# Clode the model after this reponse

resp2 = client.chat.completions.create(
    model=model_name,
    messages=[{"role": "user", "content": "Now give me a second sentence."}],
    extra_body={"keep_alive": 0},
)

print(resp2.choices[0].message.content)
```

## Method using ollama

```python
import ollama

client = ollama.Client(host='http://localhost:11434')
```

```python
resp1 = client.chat(
    model=model_name,
    messages=[{'role': 'user', 'content': 'Write one sentence about Devon.'}],
    keep_alive='30m'
)

print(resp1['message']['content'])
```

```python
resp2 = client.chat(
    model=model_name,
    messages=[{'role': 'user', 'content': 'Now write a second sentence.'}],
    keep_alive='30m'
)

print(resp2['message']['content'])
```
